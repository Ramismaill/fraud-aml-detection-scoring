```python
# 01_download — Cell 1: list files available in both Kaggle datasets (no download yet)
import sys, shutil, subprocess
sys.path.insert(0, "..")
from src.config import PATHS, Timer

BAF_REF = "sgpjesus/bank-account-fraud-dataset-neurips-2022"
AML_REF = "ealtman2019/ibm-transactions-for-anti-money-laundering-aml"

print("kaggle CLI:", shutil.which("kaggle"))

def kaggle(*args):
    """Run a kaggle CLI command and return its stdout as text."""
    r = subprocess.run(["kaggle", *args], capture_output=True, text=True)
    if r.returncode != 0:
        print("STDERR:", r.stderr)
    return r.stdout

print("\n=== BAF files ===")
print(kaggle("datasets", "files", "-d", BAF_REF))
print("=== AMLworld files ===")
print(kaggle("datasets", "files", "-d", AML_REF))
```

```python
# 01_download — Cell 2: download only the pilot files into D:/m1/raw (raw = never edited)
import zipfile

RAW_BAF = PATHS["raw"] / "baf"
RAW_AML = PATHS["raw"] / "amlworld"
RAW_BAF.mkdir(exist_ok=True)
RAW_AML.mkdir(exist_ok=True)

TARGETS = [
    (BAF_REF, "Base.csv",              RAW_BAF),
    (AML_REF, "HI-Small_Trans.csv",    RAW_AML),
    (AML_REF, "HI-Small_Patterns.txt", RAW_AML),
    (AML_REF, "HI-Small_accounts.csv", RAW_AML),
]

for ref, fname, dest in TARGETS:
    with Timer(f"download {fname}"):
        out = kaggle("datasets", "download", "-d", ref, "-f", fname, "-p", str(dest))
        print(out.strip()[-200:])          # tail of the kaggle CLI output

# Kaggle may deliver single files as .zip: extract in place, then delete the zip
for dest in (RAW_BAF, RAW_AML):
    for z in dest.glob("*.zip"):
        with zipfile.ZipFile(z) as zf:
            zf.extractall(dest)
        z.unlink()

print("\n=== files on disk ===")
for dest in (RAW_BAF, RAW_AML):
    for f in sorted(dest.iterdir()):
        print(f"{f.name:28s} {f.stat().st_size/1e6:9.1f} MB")
```
